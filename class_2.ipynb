# pattern solve


```python
n=5
for i in range(1, n+1):
    print("* " * i)
```

```python
#fibonacci
a, b = 0, 1
n = int(input("Enter the number: "))

print((f"{a}, {b}"), end="")

for i in range (1, n+1):
  next= a+b
  print(next, end="")
  if(i!=n):
    print(", ",end="")
  a,b=b,next
```
