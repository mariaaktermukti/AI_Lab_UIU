```python
import math
from queue import PriorityQueue
```

```python
def euclidean_distance(current, goal):
    """
    Calculate the Euclidean distance between the current sector and the goal.
    This is the heuristic (h(n)) used in A* to estimate the cost to the goal.
    """
    x1, y1 = current
    x2, y2 = goal

    return math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)
```

```python
def get_neighbors(grid, current):
    """
    Return a list of valid neighboring sectors (8 directions).
    Excludes obstacles (marked as -1) and sectors outside the grid.
    """
    x, y = current
    n = len(grid)

    directions = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1),    # Right
        (-1, -1),  # Up-left
        (-1, 1),   # Up-right
        (1, -1),   # Down-left
        (1, 1)     # Down-right
    ]

    neighbors = []

    for dx, dy in directions:

        nx = x + dx
        ny = y + dy

        #boundary cehck
        if 0 <= nx < n and 0 <= ny < n:

            # Check obstacle
            if grid[nx][ny] != -1:
                neighbors.append((nx, ny))

    return neighbors
```

```python
def reconstruct_path(parent, start, goal):
    """
    Reconstruct the path from start to goal using the parent dictionary.
    """
    path = []

    current = goal

    while current != start:
        path.append(current)
        current = parent[current]

    # Add start node
    path.append(start)

    # Reverse path
    path.reverse()

    return path
```

```python
def a_star_search(grid, start, goal):
    """
    Implement A* algorithm using PriorityQueue to find the shortest path.
    Returns total cost and path if found; otherwise, returns None and an empty list.
    """
    pq = PriorityQueue()

    start_g = 0
    start_f = start_g + euclidean_distance(start, goal)

    pq.put((start_f, start_g, start))

    g_score = {
        start: 0
    }

    # Parent of each node
    parent = {}

    while not pq.empty():

        # Get node with smallest f-score
        f, current_g, current = pq.get()

        if current_g != g_score.get(current, float('inf')):
            continue

        if current == goal:

            path = reconstruct_path(
                parent,
                start,
                goal
            )

            return current_g, path

       #neigh explore
        for neighbor in get_neighbors(grid, current):

            new_g = (
                current_g
                + grid[neighbor[0]][neighbor[1]]
            )

            if (
                neighbor not in g_score
                or new_g < g_score[neighbor]
            ):

                # Update g-score
                g_score[neighbor] = new_g

                # Calculate heuristic
                h = euclidean_distance(
                    neighbor,
                    goal
                )

                # Calculate f-score
                f = new_g + h

                # Store parent
                parent[neighbor] = current

                # Add to priority queue
                pq.put(
                    (f, new_g, neighbor)
                )

    # No path found
    return None, []
```

```python
def main():
    """
    Set up the grid, run A* search, and display results.
    Uses a hardcoded grid and start/goal for demonstration.
    """
    n = int(input("Enter grid size N: "))

    grid = []
    print("Enter the grid:")

    for i in range(n):
        row = list(map(int, input().split()))
        grid.append(row)

    # Start position
    Sx, Sy = map(
        int,
        input("Enter start position (Sx Sy): ").split()
    )

    # Destination position
    Dx, Dy = map(
        int,
        input("Enter destination position (Dx Dy): ").split()
    )

    start = (Sx, Sy)
    goal = (Dx, Dy)

    # Run A*
    cost, path = a_star_search(
        grid,
        start,
        goal
    )

    print("\nOptimal Cost:", cost)
    print("Optimal Path:")

    if path:
        print(
            " -> ".join(
                str(node) for node in path
            )
        )
    else:
        print([])
```

```python
main()
```
